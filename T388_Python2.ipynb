```python
character = input("Enter the word of your choice: ")
# Vowels: a,e,i,o and u
for i in character:
    j = i.lower()
    if j=="a" or j=="e" or j=="i" or j=="o" or j=="u":
        continue
    else:
        print(i)
```

# Calculate the speed (Velocity) of the car

$v^2=u^2+2as$

```python
import numpy as np
def speed_calculator(u,a,s):
    v_square= (u**2) + (2*a*s)
    speed = np.sqrt(v_square)
    print("Speed is ",round(speed,2),"m/s")

initial_vel =int(input("Enter initial Velocity: "))
acceleration =int(input("Enter Accelaration of the car: "))
distance=int(input("Enter the distance between source and destination: "))
speed_calculator(initial_vel,acceleration,distance)
```

```python
# WAP  to print whether a number is an Armstrong number
```

```python
num = int(input("Enter any number to check whether Armstrong or not: "))

length = len(str(num))
result = 0
for i in str(num):
    result= result + (int(i)**length)

if result == num:
    print("This  number", num,"is an Armstrong Number !")
else:
    print("Oh!, This number is NOT an  Armstrong Number.")
```
